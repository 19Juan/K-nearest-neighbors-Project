```python
#Para transformar un json de texto a diccionario en pandas!
#Librearía JSON me sirve para transofrmar un STRING con formato JSON en un diccionario de PANDAS
import json

dic=json.loads(df['genres'][0])
```

```python
for genres in dic:
    print(genres['name'])
```

```python
#Función para transformar el json que tengo en un diccionario y que luego recorra todas las filas del df para poner los generos bien.
def generos(fila):
    lista =[]
    fila_dic = json.loads(fila)
    for genres in fila_dic:
        lista.append(genres['name'].replace(" ",""))
    return lista
```

```python
generos(df['genres'][100])
```

```python
df['genres'] = df['genres'].apply(generos)
```

```python
df
```

```python
#Ahora hago lo mismo en columna keywiords
df['keywords']=df['keywords'].apply(generos)
```

```python
df
```

```python
#Ahoora me pide traer los 3 primeros valores de la columna cast
def elenco(fila):
    lista =[]
    fila_dic = json.loads(fila)
    largo = len(fila_dic)
    if largo > 3:
        largo = 3
    for i in range(largo):
        lista.append(fila_dic[i]['name'].replace(" ",""))
    return lista
```

```python
df['cast'] = df['cast'].apply(elenco)
```

```python
df
```

```python
#Ahora veo que tiene la columna crew en la primer fila, ya que también me piden tratarla, me piden traer los directores de las peliculas
df['crew'][0]
```

```python
def director(fila):
    fila_dic = json.loads(fila)
    for elemento in fila_dic:
        if elemento['job'] == 'Director':
            return elemento['name']
    return None
    
```

```python
df['crew'] = df['crew'].apply(director)
df
```

```python
#Ahora armaré una columna Tags, que es la suma de todas las anteriormente tratadas
df['tags'] = df['overview']+str(df['genres'])+str(df['keywords'])+str(df['cast'])+df['crew']
df.tags
```

```python
df = df.dropna()
```

```python
#Ahora se debe vectorizar

from sklearn.feature_extraction.text import TfidfVectorizer
#Una matriz tf y df me transforma cada conjunto de palabras en una matriz
#Las columnas tienen el peso relativo de la palabra en la frase (cuanto se repite la misma palabra dentro de la frase)
#luego va a tener el peso relativo en todas la frases de la columna.
from sklearn.feature_extraction.text import TfidfVectorizer

vector = TfidfVectorizer()

matriz_tfidf = vector.fit_transform(df['tags'])
```

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(n_neighbors=6,metric='cosine')
model.fit(matriz_tfidf)
```

```python
def peliculas_similares(titulo_de_pelicula):
    indice = df[df['title'] == titulo_de_pelicula].index[0]
    distances, indices = model.kneighbors(matriz_tfidf[indice])
    peliculas_similares = [(df['title'][i]) for _,i in enumerate(indices[0])]
    return peliculas_similares[1:]
```

```python
peliculas_similares('''Pirates of the Caribbean: At World's End''')
```
